Questo dominio riflette la refrattarietà terapeutica dei pazienti, definita dal fallimento di precedenti trattamenti preventivi. Il numero e il tipo di classi farmacologiche fallite rappresentano un indicatore della complessità clinica e della difficoltà di gestione dell’emicrania, ed è un potenziale predittore di risposta subottimale ai nuovi trattamenti.
- Bbloc → beta-bloccanti
- Caant → calcio-antagonisti
- Tricyclic → antidepressivi triciclici
- Antiepil → antiepilettici
- SSRISNRI → antidepressivi SSRI/SNRI
- Antiipnt → anti-ipertensivi
- Botulin → tossina botulinica

```python
refractory_vars = [
    "Bbloc", "Caant", "Tricyclic", "Antiepil",
    "SSRISNRI", "Antiipnt", "Botulin"
]

failure_rates = {}

for var in refractory_vars:
    failure_rates[var] = (
        df_baseline[var]
        .value_counts()
        .get(1, 0)
    )

pd.Series(failure_rates).sort_values(ascending=False)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

failure_df = pd.Series(failure_rates).reset_index()
failure_df.columns = ["Treatment_class", "Failure_percentage"]

sns.barplot(
    data=failure_df.sort_values("Failure_percentage", ascending=False),
    x="Failure_percentage",
    y="Treatment_class"
)
plt.title("Failure of previous preventive treatments")
plt.xlabel("Percentage of patients (%)")
plt.ylabel("Treatment class")
plt.show()
```

Therapeutic refractoriness was high, with a median of four failed preventive treatment classes per patient. While a subset of patients had no previous treatment failures, a substantial proportion failed multiple classes, highlighting marked inter-individual variability and overall clinical complexity.

```python
refractory_binary = df_baseline[refractory_vars].replace({1: 1, 2: 0})

df_baseline["N_FAILED_TREATMENTS"] = refractory_binary.sum(axis=1)

df_baseline["N_FAILED_TREATMENTS"].describe()
```

## ***Do baseline MMDs progressively decrease over time?***
> Baseline monthly migraine days progressively decreased over time, with a marked reduction during the early months of treatment followed by a sustained plateau across subsequent cycles. In the first cycle, MMDs declined sharply by month 3 and then decreased more gradually up to month 12. In cycles 2 and 3, MMDs started from lower levels and showed further modest reductions, remaining consistently below first-cycle values. A similar temporal pattern was observed for headache intensity and migraine-related disability, indicating parallel improvements in frequency, severity, and functional impact.

```python
# Teniamo solo il baseline MMDs
df_b = df_baseline[["SUBJECT_ID", "GGCEF_T0"]].copy()

# Merge nel longitudinale
df_long = df_longitudinal.merge(df_b, on="SUBJECT_ID", how="left")

# Assicuriamoci che le variabili chiave siano numeriche
cols_num = [
    "CYCLE", "MONTH", "NMDs", "GGCEF_T0",
    "INT", "HIT-6", "MIDAS", "RED_MMD_VST01"
]
for c in cols_num:
    if c in df_long.columns:
        df_long[c] = pd.to_numeric(df_long[c], errors="coerce")

df_long = df_long.sort_values(["SUBJECT_ID", "CYCLE", "MONTH"])
```

```python
mean_nmds = (
    df_long
    .groupby(["CYCLE", "MONTH"], as_index=False)["MMDs"]
    .agg(mean="mean", n="count")
)

plt.figure(figsize=(7,4))
for c, g in mean_nmds.groupby("CYCLE"):
    plt.plot(g["MONTH"], g["mean"], marker="o", label=f"CYCLE {int(c)}")

plt.xlabel("Month within cycle")
plt.ylabel("Mean NMDs")
plt.title("Mean monthly migraine days over time by cycle")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
for var, title in [
    ("INT", "Headache intensity"),
    ("HIT-6", "Headache impact (HIT-6)"),
    ("MIDAS", "Migraine disability (MIDAS)")
]:
    if var in df_long.columns:
        tmp = (
            df_long
            .groupby(["CYCLE", "MONTH"], as_index=False)[var]
            .mean()
        )

        plt.figure(figsize=(7,4))
        for c, g in tmp.groupby("CYCLE"):
            plt.plot(g["MONTH"], g[var], marker="o", label=f"Cycle {int(c)}")
        plt.title(f"{title} over time by cycle")
        plt.xlabel("Month within cycle")
        plt.ylabel(var)
        plt.legend()
        plt.tight_layout()
        plt.show()
```

## ***Are treatment effects consistent across cycles?***
> Yes — treatment effects appear consistent across cycles, with a trend toward greater reductions in later cycles.
From the end-of-cycle distributions, the median percentage reduction in monthly migraine days is substantial in all cycles and increases progressively from Cycle 1 to Cycle 3 (median ≈57% in Cycle 1, ≈67% in Cycle 2, and ≈70% in Cycle 3). Mean reductions show the same pattern. The boxplots indicate overlapping interquartile ranges across cycles, supporting overall consistency of the treatment effect, while the upward shift of medians suggests maintained—and possibly enhanced—benefit among patients remaining on treatment.
Caveat: later cycles include fewer patients due to attrition, and outliers (including negative reductions) persist, indicating heterogeneous responses. Thus, these findings are exploratory and should be confirmed with longitudinal models accounting for dropout.

```python
end_cycle = (
    df_long
    .dropna(subset=["MMDs"])
    .sort_values(["SUBJECT_ID", "CYCLE", "MONTH"])
    .groupby(["SUBJECT_ID", "CYCLE"], as_index=False)
    .tail(1)
    .rename(columns={"MMDs": "NMDs_end"})
)

end_cycle["abs_reduction"] = end_cycle["GGCEF_T0"] - end_cycle["NMDs_end"]
end_cycle["pct_reduction"] = (
    end_cycle["abs_reduction"] / end_cycle["GGCEF_T0"] * 100
)
```

```python
summary_cycle = (
    end_cycle
    .groupby("CYCLE", as_index=False)
    .agg(
        n_patients=("SUBJECT_ID", "nunique"),
        mean_pct_red=("pct_reduction", "mean"),
        median_pct_red=("pct_reduction", "median")
    )
)

summary_cycle
```

```python
plt.figure(figsize=(6,4))
plt.boxplot(
    [end_cycle.loc[end_cycle["CYCLE"]==c, "pct_reduction"].dropna()
     for c in sorted(end_cycle["CYCLE"].unique())],
    labels=[f"Cycle {int(c)}" for c in sorted(end_cycle["CYCLE"].unique())]
)
plt.ylabel("% reduction in MMDs vs baseline")
plt.title("End-of-cycle treatment effect by cycle")
plt.tight_layout()
plt.show()
```

## ***What proportion of patients achieve ≥30% and ≥50% reduction in MMDs at the end of each cycle and across the full treatment period?***
> Responder rates increased progressively across treatment cycles. At the end of Cycle 1, approximately 74% of patients achieved a ≥30% reduction in monthly migraine days and 58% achieved a ≥50% reduction. These proportions rose in Cycle 2 (≈86% and ≈71%, respectively) and further increased in Cycle 3 (≈92% and ≈82%). When considering the overall treatment period, about 72% of patients achieved a ≥30% reduction and 60% achieved a ≥50% reduction at their last available observation. These findings suggest a sustained and increasing proportion of clinically meaningful responders over time, although estimates in later cycles are influenced by patient attrition and should be interpreted as exploratory.

```python
end_cycle["resp30"] = (end_cycle["pct_reduction"] >= 30).astype(int)
end_cycle["resp50"] = (end_cycle["pct_reduction"] >= 50).astype(int)

resp_cycle = (
    end_cycle
    .groupby("CYCLE", as_index=False)
    .agg(
        n=("SUBJECT_ID", "nunique"),
        resp30_rate=("resp30", "mean"),
        resp50_rate=("resp50", "mean")
    )
)

resp_cycle["resp30_rate"] *= 100
resp_cycle["resp50_rate"] *= 100
resp_cycle
```

```python
import numpy as np

# Define ID and TIME variables as they are used globally in the spaghetti function
ID = "SUBJECT_ID"
TIME = "CYCLE"

def spaghetti(df, y, n_patients=40, seed=0):
    pts = df[ID].unique()
    np.random.seed(seed)
    sample = np.random.choice(pts, size=min(n_patients, len(pts)), replace=False)
    sub = df[df[ID].isin(sample)].dropna(subset=[TIME, y])

    plt.figure()
    for pid, dfp in sub.groupby(ID):
        plt.plot(dfp[TIME], dfp[y], alpha=0.4)
    plt.title(f"Traiettorie individuali (subset) - {y}")
    plt.xlabel("Ciclo")
    plt.ylabel(y)
    plt.show()

for y in ["MMDs", "INT", "MIDAS"]:
    if y in df_longitudinal.columns:
        spaghetti(df_longitudinal, y)
```

**MICE (Multiple Imputation) con IterativeImputer → 5 dataset imputati**

```python
from sklearn.experimental import enable_iterative_imputer  # noqa
from sklearn.impute import IterativeImputer

def mice_python(df_in, m=5, max_iter=10, seed=42, order_by_missing=True):
    """
    Crea m dataset imputati con MICE model-based.
    order_by_missing=True: euristica tipo visitSequence (prima colonne con meno NA).
    """
    df_work = df_in.copy()

    if order_by_missing:
        col_order = df_work.isna().mean().sort_values().index
        df_work = df_work[col_order]

    imputations = []
    for k in range(m):
        imp = IterativeImputer(
            max_iter=max_iter,
            random_state=seed + k,
            sample_posterior=True,  # fondamentale: introduce variabilità tra imputazioni
            skip_complete=True
        )
        arr = imp.fit_transform(df_work)
        df_imp = pd.DataFrame(arr, columns=df_work.columns, index=df_work.index)
        imputations.append(df_imp)

    return imputations

imps = mice_python(df_enc, m=5, max_iter=10, seed=42, order_by_missing=True)

print("Numero imputazioni:", len(imps))
print("Shape imputato[0]:", imps[0].shape)
print("NA nel primo imputato:", int(imps[0].isna().sum().sum()))
```

This output confirms that, using a model-based Multiple Imputation by Chained Equations (MICE) approach, we generated 5 imputed versions of the baseline dataset.

Each imputed dataset preserved the full sample size (n = 179) and contained no remaining missing values. The increase in the number of variables (55 columns) reflects the one-hot encoding of categorical predictors. This strategy allowed us to retain all subjects while accounting for uncertainty due to missing data.

**`pool()`**

Rubin’s rules

```python
def rubins_rules(fits):
    coefs = np.vstack([f.params.values for f in fits])
    vars_ = np.vstack([np.diag(f.cov_params()) for f in fits])

    Q_bar = coefs.mean(axis=0)                     # coeff medi
    U_bar = vars_.mean(axis=0)                     # var intra-imputazione
    B     = coefs.var(axis=0, ddof=1)              # var tra imputazioni

    m = len(fits)
    T = U_bar + (1 + 1/m) * B                      # var totale
    se = np.sqrt(T)

    lambda_ = ((1 + 1/m) * B) / T                  # frazione dovuta ai missing

    pooled = pd.DataFrame(
        {"coef": Q_bar, "SE": se, "lambda": lambda_},
        index=fits[0].params.index
    )
    return pooled

pooled = rubins_rules(fits)
display(pooled.head(20))
```

## IMPUTAZIONE DATI LONGITUDINALI: LOCF vs NOCB vs interpolation

**Goal**: valutare e confrontare strategie di imputazione “time-aware” per l’outcome longitudinale (MMDs), in presenza di missingness dipendente dal tempo/dropout, e scegliere il metodo più coerente con l’andamento clinico e con il pattern dei dati.

```python
# PREPARAZIONE

# Copia di lavoro
df_long = df_longitudinal.copy()

# Controllo colonne minime necessarie
needed = {"SUBJECT_ID", "CYCLE", "MONTH", "MMDs"}
missing_cols = needed - set(df_long.columns) # calcola quali colonne essenziali sono eventualmente mancanti nel dataframe
assert len(missing_cols) == 0, f"Mancano colonne nel long: {missing_cols}"

# Ordiniamo (fondamentale per operazioni temporali tipo forward/backward fill)
df_long = df_long.sort_values(["SUBJECT_ID", "CYCLE", "MONTH"])

print("Shape long:", df_long.shape)
df_long.head()
```

```python
# WIDE FORMAT
# Ogni colonna = un timepoint (Cycle x Month)

wide = df_long.pivot_table(
    index="SUBJECT_ID",
    columns=["CYCLE", "MONTH"],
    values="MMDs",
    aggfunc="mean"
)

# Rinomino colonne in modo leggibile: MMDs_C{c}_M{m}
wide.columns = [f"MMDs_C{c}_M{m}" for (c, m) in wide.columns]
wide = wide.sort_index()

print("Shape wide:", wide.shape)
print("Missing totali (wide):", int(wide.isna().sum().sum()))
wide.head()
```

```python
# ORGANIZZIAMO LE COLONNE PER CICLO E MESE
# Serve per applicare imputazione "dentro ciascun ciclo"

def parse_col(colname):
    # colname tipo "MMDs_C1_M3"
    # ritorna (cycle, month)
    c = int(colname.split("_C")[1].split("_M")[0])
    m = int(colname.split("_M")[1])
    return c, m

cols_info = [(c, m, col) for col in wide.columns for (c, m) in [parse_col(col)]]

# Dizionario: ciclo -> lista colonne ordinate per mese
cycle_cols = {}
for c in sorted(set([x[0] for x in cols_info])):
    cols_c = [(m, col) for (cc, m, col) in cols_info if cc == c]
    cols_c_sorted = [col for (m, col) in sorted(cols_c, key=lambda x: x[0])]
    cycle_cols[c] = cols_c_sorted

cycle_cols  # utile da ispezionare
```

```python
# FUNZIONI DI IMPUTAZIONE LONGITUDINALE

def impute_locf_within_cycle(wide_df, cycle_cols):
    """
    LOCF (Last Observation Carried Forward) dentro ciascun ciclo:
    per ogni soggetto, riempie NA usando l'ultimo valore osservato precedente (nel tempo).
    """
    out = wide_df.copy()
    for c, cols in cycle_cols.items():
        out[cols] = out[cols].ffill(axis=1)
    return out

def impute_nocb_within_cycle(wide_df, cycle_cols):
    """
    NOCB (Next Observation Carried Backward) dentro ciascun ciclo:
    riempie NA usando il prossimo valore osservato successivo (nel tempo).
    """
    out = wide_df.copy()
    for c, cols in cycle_cols.items():
        out[cols] = out[cols].bfill(axis=1)
    return out

def impute_interpolate_within_cycle(wide_df, cycle_cols, limit_area="inside"):
    """
    Interpolazione lineare dentro ciascun ciclo:
    - limit_area="inside": interpola SOLO buchi interni tra due osservazioni (scelta tipica per non "inventare" valori oltre l'ultima osservazione).
    """
    out = wide_df.copy()
    for c, cols in cycle_cols.items():
        # Interpola lungo l'asse del tempo (colonne ordinate per mese)
        out[cols] = out[cols].interpolate(
            axis=1, method="linear", limit_area=limit_area
        )
    return out
```

```python
# APPLICAZIONE DEI 3 METODI (COME "simple vs MICE", ma per long)

wide_locf = impute_locf_within_cycle(wide, cycle_cols)
wide_nocb = impute_nocb_within_cycle(wide, cycle_cols)
wide_intp = impute_interpolate_within_cycle(wide, cycle_cols, limit_area="inside")

print("Missing dopo LOCF:", int(wide_locf.isna().sum().sum()))
print("Missing dopo NOCB:", int(wide_nocb.isna().sum().sum()))
print("Missing dopo Interp (inside):", int(wide_intp.isna().sum().sum()))

# Interpretazione attesa:
# - LOCF e NOCB spesso riempiono quasi tutto (a meno di cicli completamente vuoti)
# - Interpolazione 'inside' NON riempie dropout terminale (è voluto)
```

Comparison of time-aware imputation strategies showed that linear interpolation did not impute any values, indicating the absence of isolated internal missing points. NOCB resulted in minimal imputation, reflecting the predominantly terminal nature of missingness. In contrast, LOCF was able to partially recover missing values in later timepoints within cycles, consistent with a dropout-driven missingness pattern.

```python
# DIAGNOSTICA 1: QUANTO STAI "IMPUTANDO" DAVVERO?
# (quanti valori erano NA e diventano non-NA)

def imputation_coverage(original, imputed):
    was_na = original.isna()
    now_filled = was_na & imputed.notna()
    coverage = now_filled.sum().sum()
    total_missing = was_na.sum().sum()
    return int(coverage), int(total_missing), float(coverage / total_missing) if total_missing > 0 else np.nan

for name, w in [("LOCF", wide_locf), ("NOCB", wide_nocb), ("INTERP", wide_intp)]:
    cov, tot, frac = imputation_coverage(wide, w)
    print(f"{name}: riempiti {cov}/{tot} missing ({frac*100:.1f}%)")
```

```python
# DIAGNOSTICA 2: TRAIETTORIE MEDIE NEL TEMPO
# Confronto: Observed-only vs LOCF vs NOCB vs Interp

def long_from_wide(wide_df, method_name):
    """
    Converte wide -> long, ricreando CYCLE e MONTH dalle colonne.
    Serve per fare plot e summary nel formato long.
    """
    tmp = wide_df.copy()
    tmp["SUBJECT_ID"] = tmp.index
    long = tmp.melt(id_vars="SUBJECT_ID", var_name="timepoint", value_name="MMDs")
    long["method"] = method_name
    # Parse cycle/month
    long["CYCLE"] = long["timepoint"].str.extract(r"_C(\d+)_").astype(int)
    long["MONTH"] = long["timepoint"].str.extract(r"_M(\d+)").astype(int)
    return long.drop(columns=["timepoint"])

# Observed-only: ricostruisco long dai valori originali (wide) senza imputazione
long_obs = long_from_wide(wide, "Observed (con NA)")
long_locf = long_from_wide(wide_locf, "LOCF")
long_nocb = long_from_wide(wide_nocb, "NOCB")
long_intp = long_from_wide(wide_intp, "Interpolation (inside)")

long_all = pd.concat([long_obs, long_locf, long_nocb, long_intp], ignore_index=True)

# Media per metodo/ciclo/mese (ignorando i NA presenti in Observed e (eventualmente) nell'interp)
mean_traj = (
    long_all
    .groupby(["method", "CYCLE", "MONTH"])["MMDs"]
    .mean()
    .reset_index()
    .sort_values(["CYCLE", "MONTH"])
)

# Plot: una figura per ciclo (leggibile)
for c in sorted(mean_traj["CYCLE"].unique()):
    plt.figure(figsize=(7,4))
    sub = mean_traj[mean_traj["CYCLE"] == c]
    for method in sub["method"].unique():
        s = sub[sub["method"] == method].sort_values("MONTH")
        plt.plot(s["MONTH"], s["MMDs"], marker="o", label=method)
    plt.title(f"Mean MMDs trajectory – Cycle {c}")
    plt.xlabel("Month")
    plt.ylabel("Mean MMDs")
    plt.grid(True)
    plt.legend()
    plt.show()

# Come leggere questi plot (stile lab):
# - LOCF tende ad "appiattire" le code (dropout terminale)
# - NOCB può "spostare indietro" info dal futuro (attenzione interpretativa)
# - Interpolazione 'inside' preserva i buchi terminali (più conservativa sul dropout)
```

**Traiettorie medie per ciclo**

- Observed vs LOCF:

  - LOCF segue bene l’andamento osservato nei primi mesi

  - poi si appiattisce → tipico effetto LOCF sul dropout

- NOCB:

  - curve “strane”, spesso sopra/sotto l’osservato

  - introduce informazione dal futuro

- Interpolation:

  - praticamente sovrapposta all’Observed perché non imputa nulla

**Conclusione**:
il missing non è “interno”, è dropout terminale → interpolazione non applicabile

```python
# DIAGNOSTICA 3: DISTRIBUZIONE DEI VALORI IMPUTATI (solo dove prima era NA)
# Confronto tra metodi: cosa stai "inventando"?

def imputed_values_only(original, imputed):
    """
    Estrae solo i valori imputati: posizioni che erano NA e ora sono non-NA.
    """
    mask = original.isna() & imputed.notna()
    vals = imputed.where(mask).stack().values  # array 1D
    return vals

vals_locf = imputed_values_only(wide, wide_locf)
vals_nocb = imputed_values_only(wide, wide_nocb)
vals_intp = imputed_values_only(wide, wide_intp)

plt.figure(figsize=(7,4))
plt.hist(vals_locf, bins=30, alpha=0.5, density=True, label="LOCF (imputed only)")
plt.hist(vals_nocb, bins=30, alpha=0.5, density=True, label="NOCB (imputed only)")
plt.hist(vals_intp, bins=30, alpha=0.5, density=True, label="Interpolation (imputed only)")
plt.title("Distribution of imputed MMDs values (only formerly-missing cells)")
plt.xlabel("MMDs")
plt.ylabel("Density")
plt.grid(True)
plt.legend()
plt.show()

# Interpretazione:
# - Se una distribuzione è troppo "a spike" -> metodo che replica valori (LOCF/NOCB)
# - Se interpolazione produce valori più continui -> coerente, ma solo per buchi interni
```

**Distribuzione dei valori imputati (solo celle prima NA)**

```python
base = df_baseline.copy()

# ---- Ricostruisci feature derivate se non esistono già ----
# failed treatments
refractory_vars = ["Bbloc", "Caant", "Tricyclic", "Antiepil", "SSRISNRI", "Antiipnt", "Botulin"]
refractory_vars = [v for v in refractory_vars if v in base.columns]
if "N_FAILED_TREATMENTS" not in base.columns and len(refractory_vars) > 0:
    refractory_binary = base[refractory_vars].replace({1: 1, 2: 0})
    base["N_FAILED_TREATMENTS"] = refractory_binary.sum(axis=1)

# comorbidities
comorbidities = ["Psycopathological", "Hypertension", "Sleep_Disorders"]
comorbidities = [v for v in comorbidities if v in base.columns]
if "N_COMORBIDITIES" not in base.columns and len(comorbidities) > 0:
    comorb_binary = base[comorbidities].replace({1: 1, 2: 0})
    base["N_COMORBIDITIES"] = comorb_binary.sum(axis=1)

# ---- Selezione covariate core ----
core_vars = [ID, "AGE", "SEX", "DIAGNOSIS", "ANTIBODY", "GGCEF_T0", "INT_T0",
             "N_FAILED_TREATMENTS", "N_COMORBIDITIES", "Suspension", "MONTHS_OF_TREAT"]

core_vars = [v for v in core_vars if v in base.columns]
base_core = base[core_vars].copy()

print("Variabili baseline usate:", core_vars)
print("Missing totali in base_core:", int(base_core.isna().sum().sum()))
base_core.head()
```

**Imputazione baseline** (stile MICE) su covariate core

- one-hot encoding per categoriche (SEX, DIAGNOSIS, ANTIBODY)

- IterativeImputer con sample_posterior=True per creare m imputazioni

- fit modelli su ciascuna imputazione e pool (Rubin).

```python
# 1) separa ID
base_core = base_core.set_index(ID)

# 2) identifica categoriche (qui sono codificate numericamente ma concettualmente categoriche)
cat_like = [c for c in ["SEX", "DIAGNOSIS", "ANTIBODY"] if c in base_core.columns]

# 3) one-hot encoding (drop_first=True per evitare multicollinearità)
base_enc = pd.get_dummies(base_core, columns=cat_like, drop_first=True)

print("Shape baseline encoded:", base_enc.shape)
print("Missing totali encoded:", int(base_enc.isna().sum().sum()))

def mice_like_imputations(df_in, m=5, max_iter=20, seed=42):
    """Crea m dataset imputati con IterativeImputer (stile MICE)."""
    imps = []
    for k in range(m):
        imp = IterativeImputer(
            max_iter=max_iter,
            random_state=seed + k,
            sample_posterior=True,
            skip_complete=True
        )
        arr = imp.fit_transform(df_in)
        df_imp = pd.DataFrame(arr, index=df_in.index, columns=df_in.columns)
        imps.append(df_imp)
    return imps

m = 5
imps_base = mice_like_imputations(base_enc, m=m, max_iter=20, seed=42)

print("Imputazioni create:", len(imps_base))
print("NA nella prima imputazione:", int(imps_base[0].isna().sum().sum()))
```

## 2.3: Mixed-effects model (primaria) su observed-only

**Goal**

Stimare l’effetto del tempo e dei cicli su MMDs, aggiustando per baseline.

**Modello**:

- **outcome**: MMDs

- **fixed effects**: time_abs, CYCLE (categorico), interazione time_abs*CYCLE

- **covariate baseline**: GGCEF_T0, INT_T0, + altre core (failed, comorb, sesso, diagnosi, antibody)

- random intercept per paziente: (1 | SUBJECT_ID)

In statsmodels useremo MixedLM.from_formula.

```python
def fit_mixed_model(df_long, df_base_imp, use_interaction=True):
    """
    Fit LMM: MMDs ~ time_abs + C(CYCLE) + time_abs:C(CYCLE) + baseline covariates
    Random intercept per SUBJECT_ID.
    """
    # merge (baseline imputata) nel longitudinal
    dat = df_long.merge(df_base_imp.reset_index(), on=ID, how="left")

    # outcome non missing
    dat = dat.dropna(subset=["MMDs", "time_abs", "CYCLE"]).copy()

    # formula: costruzione dinamica
    fixed_terms = ["time_abs", "C(CYCLE)"]
    if use_interaction:
        fixed_terms.append("time_abs:C(CYCLE)")

    # aggiungi covariate baseline se esistono
    base_covs = [c for c in df_base_imp.columns if c not in ["Suspension", "MONTHS_OF_TREAT"]]
    # (df_base_imp qui è già encoded, quindi include GGCEF_T0, INT_T0, N_FAILED_TREATMENTS, dummies ecc.)
    fixed_terms += base_covs

    # rimuovi duplicati mantenendo ordine
    fixed_terms_unique = []
    for t in fixed_terms:
        if t not in fixed_terms_unique:
            fixed_terms_unique.append(t)

    formula = "MMDs ~ " + " + ".join(fixed_terms_unique)

    # Fit MixedLM
    model = MixedLM.from_formula(
        formula=formula,
        groups=dat[ID],
        data=dat
    )
    res = model.fit(reml=False, method="lbfgs")
    return res

# Fit per ciascuna imputazione baseline
fits_lmm = []
for k, bimp in enumerate(imps_base, start=1):
    try:
        res = fit_mixed_model(long_obs_primary, bimp, use_interaction=True)
        fits_lmm.append(res)
        print(f"[OK] Fit LMM imputazione {k}/{m}")
    except Exception as e:
        print(f"[FAIL] Fit LMM imputazione {k}/{m}: {e}")

print("Modelli LMM stimati:", len(fits_lmm))
```

```python
import numpy as np
import pandas as pd
from scipy.stats import linregress

df_final   # long, con MMDs imputate via LOCF
"""
trasformato ogni paziente da:
una serie temporale (molte righe) a un vettore di caratteristiche che descrive
la sua traiettoria.
Per ogni paziente e per ogni ciclo ho calcolato:
- media delle MMDs → livello medio di burden
- deviazione standard → stabilità / fluttuazione
- slope → velocità e direzione del cambiamento
Il clustering:
non funziona bene su time series grezze
funziona bene su rappresentazioni compatte e comparabili
"""
features = []

for pid, g in df_final.groupby("SUBJECT_ID"):
    row = {"SUBJECT_ID": pid}

    for c in sorted(g["CYCLE"].unique()):
        gc = g[g["CYCLE"] == c]

        if gc.shape[0] >= 2:
            # mean
            row[f"C{c}_mean"] = gc["MMDs"].mean()
            # sd
            row[f"C{c}_sd"] = gc["MMDs"].std()
            # slope
            slope, _, _, _, _ = linregress(gc["MONTH"], gc["MMDs"])
            row[f"C{c}_slope"] = slope
        else:
            row[f"C{c}_mean"] = np.nan
            row[f"C{c}_sd"] = np.nan
            row[f"C{c}_slope"] = np.nan

    features.append(row)

df_feat = pd.DataFrame(features).set_index("SUBJECT_ID")
df_feat.head()
```

Le differenze che osservo tra i cluster per una certa variabile baseline sono compatibili con semplici fluttuazioni casuali, oppure sono sufficientemente sistematiche da suggerire una reale eterogeneità tra i gruppi?


```python
cat_vars = ["SEX", "DIAGNOSIS", "ANTIBODY", "Suspension"]

for v in cat_vars:
    print(f"\nVariable: {v}")
    tab = pd.crosstab(df_char["cluster"], df_char[v], normalize="index")
    display(tab)
```

Although longitudinal phenotypes were defined exclusively on the basis of MMD trajectories, they showed distinct clinical profiles. Demographic variables such as sex and diagnosis were broadly comparable across clusters, whereas treatment-related variables differed substantially. In particular, treatment discontinuation rates varied markedly across phenotypes, with poor-responder clusters exhibiting a high likelihood of suspension and responder clusters showing sustained treatment persistence. These findings support the clinical relevance of the identified longitudinal phenotypes.

```python
from scipy.stats import chi2_contingency

for v in cat_vars:
    tab = pd.crosstab(df_char["cluster"], df_char[v])
    if tab.shape[1] > 1:
        chi2, p, _, _ = chi2_contingency(tab)
        print(f"{v}: chi2 p = {p:.4f}")
```
